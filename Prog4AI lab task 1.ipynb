```python
#task1
class student:
    def __init__(self,name,age,course):
        self.name= name 
        self.age= age
        self.course= course
    def display(self):
           print("Information:")
           print("Name:",self.name)
           print("Age",self.age)
           print("Course",self.course)
        
s1=student("Rahima",19,"Programming") 
s2=student("Zara",20,"python")
s1.display()
s2.display()
```

```python
#task 2
class graduate(student):
    def __init__(self,name,age,course,research):
        super().__init__(name,age,course)
        self.research=research
    def display1(self):
           print("---Graduated Students---")
           super().display()
           print("Research Topics:",self.research)
g=graduate("Sara",21,"Python","AI")
g.display1()
        
```
